```python
# Task: "tags" sütununda ki veriler parçalanıp sayılarak "tag_count" isimli yeni bir sütuna yazdırılacak
# Path I
# def calculate_tags(tag: str) -> int:
#     return len(tag.split('|'))

# df['tag_count'] = df['tags'].apply(calculate_tags)

# df[['title', 'tags', 'tag_count']]
```

```python
# Path II
df['tag_count'] = df['tags'].apply(lambda x: len(x.split('|')))

df[['title', 'tags', 'tag_count']]
```

```python
# en yüksek tag_count sahip 10 içerik
df.sort_values(
    by='tag_count',
    ascending=False
).head(10)
```

```python
# tag_count 30'dan büyük olan içerikleri yazdırın
df.query('tag_count > 30')[['title', 'tags', 'tag_count']].sort_values(by='tag_count', ascending=False)
```

```python
# (likes / likes + dislike) formülü üzerinden hesaplanarak like_avg sütununa yazalım
def like_dislike_avg(likes: pd.Series, dislikes: pd.Series) -> list:
    like_list = list(likes)
    dislike_list = list(dislikes)

    combine_list = list(
        zip(like_list, dislike_list)
    )

    avg_list = []

    for like, dislike in combine_list:
        if like + dislike == 0:
            avg_list.append(0)
        else:
            avg = like / (like + dislike)
            avg_list.append(avg)
    
    return avg_list

df['like_avg'] = like_dislike_avg(likes=df['likes'], dislikes=df['dislikes'])

df[['title', 'likes', 'dislikes', 'like_avg']]
```

```python
# kanalların like avg nedir?
(
    df.groupby(by='channel_title')
        .agg({
            'like_avg': 'mean',
            'comment_count': 'sum',
            'views': 'sum'
        })
        .sort_values(
            by='views',
            ascending=False
        ).head(10)
)
```
