```python
df.columns =  ["symboling", "normalized-losses", "make", "fuel-type", "aspiration", "num-of-doors", "body-style",
              "drive-wheels", "engine-location", "wheel-base", "length", "width", "height", "curb-weight",
              "engine-type", "num-of-cylinders", "engine-size", "fuel-system", "bore", "stroke", "compression-ratio",
              "horsepower", "peak-rpm", "city-mpg", "highway-mpg", "price"]

df.columns
```

```python
df
```

```python
df.replace(
    to_replace='?',
    value=np.nan,
    inplace=True
)

df.head()
```

```python
df.isnull().sum()
```

```python
missing_values_df = df.isnull()

missing_values_df.head()
```

```python
missing_dict = {}

for column_name in missing_values_df.columns.tolist():
    print(
        '============================\n'
        f'Column Name: {column_name}\n'
        f'Missing Values Count: {missing_values_df[column_name].value_counts()}'
    )
```

```python
# ortalama değeri bulup missing value yerine yerleştirelim
df['normalized-losses'] = df['normalized-losses'].replace(
    to_replace=np.nan,
    value=df['normalized-losses'].astype(float).mean()
)

df.head()
```

```python
df['num-of-doors'].value_counts()
```

```python
df['num-of-doors'].value_counts().idxmax() # satır etiketini döner
```

```python
df['num-of-doors'].value_counts().argmax() # index döner
```

```python
# frekans aralığını bulup missing value yerine yerleştirelim
df['num-of-doors'] = df['num-of-doors'].replace(
    to_replace=np.nan,
    value=df['num-of-doors'].value_counts().idxmax()
)
```

# Veri Stadardizayonu

```python
df['city_l/100km'] = 235 / df['city-mpg']
df['highway_l/100km'] = 235 / df['highway-mpg']

df[["make", 'city-mpg', 'city_l/100km']]
```
