```python
import pandas as pd
```

```python
customers = {
    'customer_id': [1, 2, 3],
    'full_name': ['Burak Yılmaz', 'Hakan Yılmaz', 'İpek Yılmaz'],
    'address': ['Ataköy', 'Sarıyer', 'Sarıyer']
}

orders = {
    'order_id': [1001, 1002, 1003, 1004, 1005],
    'customer_id': [1, 2, 3, 4, 5],
    'order_revenue': [2456, 1234, 4567, 2345, 9876]
}

df_customers = pd.DataFrame(customers)
df_orders = pd.DataFrame(orders)
```

```python
df_customers
```

```python
df_orders
```

# Merge

```python
# inner merge
inner_merge = pd.merge(
    left=df_customers,
    right=df_orders,
    on='customer_id',
    #left_on='column_name',
    #right_on='column_name'
    how='inner'
)

inner_merge
```

```python
# left merge
left_merge = pd.merge(
    left=df_orders,
    right=df_customers,
    on='customer_id',
    how='left'
)

left_merge
```

```python
# right merge
right_merge = pd.merge(
    left=df_customers,
    right=df_orders,
    on='customer_id',
    how='right'
)

right_merge
```
