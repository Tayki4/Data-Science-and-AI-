# Data Fitering

```python
# 2000'den sonra yayınlanmış rating en çok olan top 10 movie
(
    df.query('YR_Released > 2000')
        [['Movie_Title', 'YR_Released','Rating']]
        .sort_values(by='Rating', ascending=False)
        .head(10)
)
```

```python
# (
#     df[df['YR_Released'] > 2000]
#         .loc[:, ['Movie_Title', 'YR_Released','Rating']]
#         .sort_values(by='Rating', ascending=False)
#         .head(10)
# )

(
    df[df['YR_Released'] > 2000]
        [['Movie_Title', 'YR_Released','Rating']]
        .sort_values(by='Rating', ascending=False)
        .head(10)
)
```

```python
# Rating'i 7.0 ve 9.0 arasında olan filmlerin 'Movie_Title','Rating', 'Num_Reviews' sütunları Rating sütununa göre çoktan aza sıralayarak getirin

(
    df[
        (df['Rating'] > 7.0) & (df['Rating'] < 9.0)
    ][['Movie_Title', 'YR_Released', 'Rating']]
    .sort_values(by='Rating', ascending=False)   
)
```

```python
# Num_Reviews 2000'den 3000'den az olan filmlerin 'Movie_Title', 'Rating', 'Num_Reviews'
# Num_Reviews verisi çoktan aza sort edelim
(
    df.query('2000 < Num_Reviews < 3000')
        .loc[: , ['Movie_Title', 'Rating', 'Num_Reviews']]
        .sort_values(
            by='Num_Reviews', 
            ascending=False
        )
)
```

```python
# Movie_Title içerisinde "King" ifadesi geçen ve Rating 7.0'den büyük olan filmlerin 'Movie_Title', 'Rating', 'Num_Reviews'
# Rating sütununa göre çoktan aza sıralayalım
(
    df.query('Movie_Title.str.contains("King") & Rating >= 7.0')
        .loc[:, ['Movie_Title', 'Rating', 'Num_Reviews']]
        .sort_values(
            by='Rating', 
            ascending=False
        )
)
```

# Group By

```python
# Filmlerin yayınlanma yıllarına göre toplam Num_Reviews sayılarını yazdıralım
# Sorunun mantığı: YR_Released sütununa göre veriyi gruplayalım, Num_Reviews sütununu toplayalım
# Path I
df.groupby(by='YR_Released')['Num_Reviews'].sum().sort_values(ascending=False)
```

```python
# Path II
(
    df.groupby(by='YR_Released')
        .agg({
            'Num_Reviews': 'sum'
        })
        .query('7000000 < Num_Reviews < 10000000')
        .sort_values(
            by='Num_Reviews', 
            ascending=False
        )
)
```

```python
# YR_Released göre veriyi gruplayalım
# Rating Ortalmaası bulalım
(
    df.groupby(by='YR_Released')
        .agg({
            'Rating': 'mean'
        })
        .query('8.0 < Rating < 10.0')
        .sort_values(
            by='Rating', 
            ascending=False
        )
)
```
