```python
import pandas as pd
```

```python
df = pd.read_csv(
    filepath_or_buffer='data/nba.csv'
)

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
# En değerli takım kim
df_groupby_salary = (
    df.groupby(by='Team')
        .agg({
            'Salary': 'sum'
        })
        .sort_values(
            by='Salary',
            ascending=False
        )
        .head(1)
)
# $ 10,698,868.90 apply() ve lambda ile yapılacak
df_groupby_salary['Salary'] = df_groupby_salary['Salary'].apply(lambda x: f'$ {x:,.2f}')
df_groupby_salary
```

```python
df_groupby_team = (
    df.groupby(by='Team')
        .agg({
            'Age': 'mean'
        })
        .sort_values(
            by='Age',
            ascending=False
        )
)

# Hint: apply() araştır ve lambda ile yapın. Age sütunundaki bilgilerin noktadan sonra 2 hanesi alınsın
df_groupby_team['Age'] = df_groupby_team['Age'].apply(lambda x: f'{x:.2f}')

df_groupby_team
```

```python
df_college = (
    df.groupby(by='College')
        .agg({
            'Salary': 'mean'
        })
        .sort_values(
            by='Salary',
            ascending=False
        )
        .head(10)
)

df_college['College'] = df_college.index
df_college.reset_index(drop=True, inplace=True)
df_college['Salary'] = df_college['Salary'].apply(lambda x: f'$ {x:,.2f}')
df_college
```

```python
# Custom Function yazalım
# Name sütunun bulunan bilgiler içerisinde "and" ifadesi geçen satırları filtereleyen bir fonksiyon yazalım
def str_find(name: str, prefix: str = 'and') -> bool:
    if prefix.lower() in name.lower():
        return True
    else:
        return False

df[
    df['Name'].apply(str_find)
]
```
